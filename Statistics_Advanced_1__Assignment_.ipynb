6. What is the Central Limit Theorem (CLT), and why is it critical in statistics?
   - The Central Limit Theorem states that when sufficiently large random samples are taken from a population, the sampling distribution of the sample mean becomes approximately normal, regardless of the original population's distribution, under common regularity conditions.

     The standard error of the sample mean is:

$$ SE=\frac{\sigma}{\sqrt{n}} $$

7. What is the significance of confidence intervals in statistical analysis?
   - A confidence interval (CI) gives a range of plausible values for an unknown population parameter, based on sample data.

     For example, if a company's average customer spending is estimated as:

      ₹500 ± ₹30

      then the interval is:

      ₹470 to ₹530

      A 95% confidence interval means that if we repeatedly took samples and constructed intervals using the same procedure, approximately 95% of those intervals would contain the true population parameter.

      Confidence intervals are useful because they show both:

      - The estimated value
      - The uncertainty/precision of the estimate

8.  What is the concept of expected value in a probability distribution?
    - The expected value represents the long-run average outcome of a random variable if the experiment is repeated many times.

      For a discrete random variable:

      $$ E(X)=\sum xP(X=x) $$

      For example, suppose a game has:

      Win ₹100 with probability 0.5
      Lose ₹50 with probability 0.5

      Then:

      $$ E(X)=(100)(0.5)+(-50)(0.5)=25 $$

      So the expected value is ₹25.

9. Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.
   - We need to generate 1,000 values with:

     Mean = 50
     Standard deviation = 5

     Then calculate their sample mean and standard deviation and visualize them using a histogram.
    
    

   

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random numbers
np.random.seed(42)

data = np.random.normal(loc=50, scale=5, size=1000)

# Calculate mean and standard deviation
mean_value = np.mean(data)
std_value = np.std(data)

print("Mean:", mean_value)
print("Standard Deviation:", std_value)

# Draw histogram
plt.hist(data, bins=30, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

10. You are working as a data analyst for a retail company. The company has
collected daily sales data for 2 years and wants you to identify the overall sales trend.
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,

● Explain how you would apply the Central Limit Theorem to estimate the average sales
with a 95% confidence interval.
● Write the Python code to compute the mean sales and its confidence interval.



```python
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
```

Applying the CLT

We can use the sample of daily sales to estimate the population's average daily sales.

Steps:

- Calculate the sample mean.
- Calculate the sample standard deviation.
- Calculate the standard error:

  $$ SE=\frac{s}{\sqrt{n}} $$

  For a 95% confidence interval, because the population standard deviation is unknown and the sample is relatively small, use the t-distribution:
  
  $$ CI=\bar{x}\pm t_{0.975,n-1}\frac{s}{\sqrt n} $$
  
  Interpret the resulting interval as a range of plausible values for the population mean.

```python
import numpy as np
from scipy import stats

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Sample size
n = len(daily_sales)

# Mean and sample standard deviation
mean_sales = np.mean(daily_sales)
std_sales = np.std(daily_sales, ddof=1)

# Standard error
standard_error = std_sales / np.sqrt(n)

# 95% confidence interval using t-distribution
confidence_level = 0.95
t_critical = stats.t.ppf((1 + confidence_level) / 2, df=n-1)

margin_of_error = t_critical * standard_error

lower = mean_sales - margin_of_error
upper = mean_sales + margin_of_error

print("Mean Sales:", mean_sales)
print("Sample Standard Deviation:", std_sales)
print("95% Confidence Interval:", (lower, upper))
```
